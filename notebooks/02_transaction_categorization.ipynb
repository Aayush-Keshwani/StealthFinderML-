# Notebook 02: Automated Transaction Categorization
### Model 1: Baseline vs XGBoost Multiclass Classification

Evaluates Dummy baseline, Logistic Regression TF-IDF baseline, and XGBoost classifier. Implements confidence thresholding with low-confidence 'Needs Review' routing.

```python
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
from stealth_spend.categorization import TransactionCategorizer
from stealth_spend.preprocessing import Preprocessor

# Load cleaned interim data
df = pd.read_csv('../reports/generated/cleaned_transactions.csv')
prep = Preprocessor()
train_df, test_df = prep.train_test_split(df, test_ratio=0.25)

categorizer = TransactionCategorizer(min_confidence_threshold=0.40)
metrics = categorizer.train(train_df, val_df=test_df)

print(f"Dummy Baseline Macro F1: {metrics['dummy_baseline']['macro_f1']:.3f}")
print(f"Logistic Regression Macro F1: {metrics['logistic_regression']['macro_f1']:.3f}")
print(f"XGBoost Macro F1: {metrics['xgboost']['macro_f1']:.3f}")
print(f"XGBoost Accuracy: {metrics['xgboost']['accuracy']:.3f}")
```
