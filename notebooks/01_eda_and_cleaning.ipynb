# Notebook 01: Exploratory Data Analysis & Statement Cleaning
### Stealth-Spend: Financial Behavior Modeler & Anomaly Detector

This notebook ingests raw bank statements, validates schema integrity and missingness, sanitizes text descriptions, and harmonizes transaction amounts per convention (debits negative, credits positive).

```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from stealth_spend.data_ingestion import normalize_to_canonical
from stealth_spend.validation import DataValidator
from stealth_spend.preprocessing import Preprocessor

# Ingest provided raw statement
raw_df = pd.read_csv('../transactions.csv')
print(f"Raw transactions loaded: {len(raw_df)}")
raw_df.head(3)
```

```python
# Normalize to canonical transaction schema
canonical_df = normalize_to_canonical(raw_df, source_filename='transactions.csv')
validator = DataValidator()
quality_report = validator.validate(canonical_df)
print("Validation Status:", quality_report['status'])
print("Total Rows:", quality_report['total_rows'])
print("Duplicate IDs:", quality_report['duplicate_ids'])

quality_summary = validator.get_quality_summary_df(canonical_df)
quality_summary
```

```python
# Preprocessing and text normalization
prep = Preprocessor()
clean_df = prep.process(canonical_df)
print("Cleaned transactions sample:")
clean_df[['transaction_date', 'description_raw', 'description_clean', 'amount', 'category']].head(5)
```
